```python
from IPython.display import Video as IPyVideo
import moviepy.editor as mpe
import numpy as np
```

## 1. Carregando o vídeo:

```python
clip = mpe.VideoFileClip("Bad_apple.mp4")
clip.ipython_display(maxduration=100, width=480)
```

## 2. Invertendo o vídeo a cada 20 segundos:

```python
def invert_frame(get_frame, t):
    frame = get_frame(t)
    
    if (t // 20) % 2 == 1:
        return np.fliplr(frame)
    
    return frame

inverted_clip = clip.fl(invert_frame)

inverted_clip.ipython_display(maxduration=100, width=480)
```

## 3. Diminuindo o volume do vídeo:

```python
duration = clip.duration

def fade_audio(audio, t):
    frame = audio(t)

    volume = np.where(
        t < duration - 10,  
        1 - (t / (duration - 10)), 
        0
    )

    frame[:, 0] *= volume
    frame[:, 1] *= volume

    return frame

audio_fadeout = inverted_clip.audio.fl(fade_audio)
video_audio_edited = inverted_clip.set_audio(audio_fadeout)

video_audio_edited.ipython_display(maxduration=100, width=480)

```

## 4. Modificando o tempo do vídeo: 

```python
first_part = video_audio_edited.subclip(0, 60)
second_part = video_audio_edited.subclip(80, video_audio_edited.duration)
cut_part = video_audio_edited.subclip(60, 80)

final_video = mpe.concatenate_videoclips([first_part, second_part, cut_part])

final_video.ipython_display(maxduration=100, width=480)
```

## 5. Salvando o vídeo:

```python
final_video.write_videofile("Bad_apple_editado.mp4")
```
